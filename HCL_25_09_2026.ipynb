```python
def longest_unique_sequence(student_ids):
    seen = set()
    left = 0
    longest = 0
    for right in range(len(student_ids)):
        while student_ids[right] in seen:
            seen.remove(student_ids[left])
            left += 1
        seen.add(student_ids[right])
        current_length = right - left + 1
        if current_length > longest:
            longest = current_length

    return longest
student_ids = list(map(int, input("Enter studentID: ").split()))
answer = longest_unique_sequence(student_ids)
print("Longest sequence without repeated studentID:", answer)
```

```python
def maximum_discount(discounts):
    current_sum = discounts[0]
    maximum_sum = discounts[0]
    for i in range(1, len(discounts)):
        # Either start a new range or continue the previous range
        current_sum = max(discounts[i], current_sum + discounts[i])
        if current_sum > maximum_sum:
            maximum_sum = current_sum
    return maximum_sum
discounts = list(map(int, input("Enter discount values: ").split()))
answer = maximum_discount(discounts)
print("Maximum total discount:", answer)
```

```python
def trapped_water(heights):
    if len(heights) < 3:
        return 0
    left = 0
    right = len(heights) - 1
    left_max = 0
    right_max = 0
    total_water = 0
    while left < right:
        if heights[left] <= heights[right]:
            if heights[left] >= left_max:
                left_max = heights[left]
            else:
                total_water += left_max - heights[left]
            left += 1

        else:
            if heights[right] >= right_max:
                right_max = heights[right]
            else:
                total_water += right_max - heights[right]

            right -= 1

    return total_water
heights = list(map(int, input("Enter building heights: ").split()))
answer = trapped_water(heights)
print("Total trapped rainwater:", answer)
```

```python
def highest_performance_period(scores):
    current_sum = scores[0]
    best_sum = scores[0]

    for i in range(1, len(scores)):
        # Decide whether to start a new period or continue
        current_sum = max(scores[i], current_sum + scores[i])

        if current_sum > best_sum:
            best_sum = current_sum

    return best_sum
scores = list(map(int, input("Enter monthly performance scores: ").split()))
answer = highest_performance_period(scores)
print("Highest overall performance:", answer)
```

```python
def maximum_product(values):
    maximum_product = values[0]
    minimum_product = values[0]
    answer = values[0]

    for i in range(1, len(values)):
        value = values[i]

        # A negative number can turn the minimum into the maximum
        if value < 0:
            maximum_product, minimum_product = minimum_product, maximum_product

        maximum_product = max(value, maximum_product * value)
        minimum_product = min(value, minimum_product * value)

        if maximum_product > answer:
            answer = maximum_product

    return answer


sales = list(map(int, input("Enter sales values: ").split()))

answer = maximum_product(sales)

print("Maximum product:", answer)
```

```python
def longest_unique_purchases(products):
    seen = set()
    left = 0
    longest = 0
    for right in range(len(products)):
        while products[right] in seen:
            seen.remove(products[left])
            left += 1

        seen.add(products[right])
        current_length = right - left + 1
        if current_length > longest:
            longest = current_length

    return longest
products = list(map(int, input("Enter product IDs: ").split()))

answer = longest_unique_purchases(products)

print("Longest sequence of unique purchases:", answer)
```

```python
def count_target_groups(transactions, target):
    prefix_sum = 0
    count = 0
    prefix_counts = {0: 1}
    for amount in transactions:
        prefix_sum += amount
        required_sum = prefix_sum - target
        if required_sum in prefix_counts:
            count += prefix_counts[required_sum]
        if prefix_sum in prefix_counts:
            prefix_counts[prefix_sum] += 1
        else:
            prefix_counts[prefix_sum] = 1

    return count
transactions = list(map(int, input("Enter transaction amounts: ").split()))
target = int(input("Enter target amount: "))
answer = count_target_groups(transactions, target)
print("Number of continuous transaction groups:", answer)
```

```python
def group_skills(skill_codes):
    groups = {}
    for skill in skill_codes:
        key = ''.join(sorted(skill))
        if key not in groups:
            groups[key] = []
        groups[key].append(skill)
    return groups
skill_codes = input("Enter skill codes separated by spaces: ").split()
groups = group_skills(skill_codes)
print("Skill groups:")
for group in groups.values():
    print(group)
```

```python
def longest_consecutive_sequence(numbers):
    numbers_set = set(numbers)
    longest = 0
    for number in numbers_set:
        if number - 1 not in numbers_set:
            current_number = number
            current_length = 1
            while current_number + 1 in numbers_set:
                current_number += 1
                current_length += 1
            if current_length > longest:
                longest = current_length

    return longest
packets = list(map(int, input("Enter packet IDs: ").split()))

answer = longest_consecutive_sequence(packets)

print("Longest consecutive sequence length:", answer)
```

```python
def merge_appointments(appointments):
    if not appointments:
        return []
    appointments.sort(key=lambda x: x[0])
    merged = [appointments[0]]
    for current in appointments[1:]:
        previous = merged[-1]
        if current[0] <= previous[1]:
            previous[1] = max(previous[1], current[1])

        else:
            merged.append(current)

    return merged


n = int(input("Enter number of appointments: "))

appointments = []

for i in range(n):
    start, end = map(int, input(
        "Enter start and end time: "
    ).split())

    appointments.append([start, end])


result = merge_appointments(appointments)

print("Merged appointment schedule:")

for appointment in result:
    print(appointment[0], appointment[1])
```
