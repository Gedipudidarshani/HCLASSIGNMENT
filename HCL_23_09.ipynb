```python
num = input().split(",")
result = []
for n1 in num:
    decimal = int(n1, 2)
    if decimal % 5 == 0:
        result.append(n1)
print(",".join(result))
```

```python
s=input()
l=0
d=0

for ch in s:
    if ch.isalpha():
        l+= 1
    elif ch.isdigit():
        d+= 1

print("LETTERS", l)
print("DIGITS", d)
```

```python
n=int(input())
fact=1
for i in range(1, n+1):
    fact *= i
print(fact)
```
